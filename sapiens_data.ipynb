```python
# a function that outputs whether a cell is cycling or not 
def scoring(cycling, non_cycling):
    if cycling == non_cycling:
        decision = 'not clear'
    elif cycling > non_cycling:
        decision = 'cycling'
    else:   
        decision = 'non_cycling'
    return(decision)
```
